```python
import pandas as pd
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
data = pd.read_csv("Cleaned_Dataset.csv")
data.head()
```

```python
data.drop(columns=['Student_ID','Student_Name'], axis=1, inplace=True)

data.head()
```

```python
data['Gender'] = data['Gender'].map({
    'Male': 0,'Female': 1
})
data.head()
```

```python
X = data.drop('Final_Grade', axis=1)
y = data['Final_Grade']

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled[:5])
```

```python
X = data.drop('Final_Grade', axis=1)
y = data['Final_Grade']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[:5])
```

```python
# Numeric (for regression)
data['Final_Grade_Numeric'] = data['Final_Grade'].map({
    'A': 90, 'B': 75, 'C': 60, 'D': 40, 'E': 30, 'F': 20
})

# Binary classification
data['Result'] = data['Final_Grade'].map({
    'A': 1, 'B': 1, 'C': 1,
    'D': 0, 'E': 0, 'F': 0
})

# Label for display
data['Result_Label'] = data['Result'].map({1: 'Pass', 0: 'Fail'})

# Multi-class label
data['Grade_Label'] = data['Final_Grade_Numeric'].apply(
    lambda x: 'A' if x >= 90 else
              'B' if x >= 75 else
              'C' if x >= 60 else
              'D' if x >= 40 else
              'F'
)
```

```python
data.head(5)
```

```python
X = data.drop(['Final_Grade', 'Result_Label', 'Grade_Label'], axis=1)
y = data['Final_Grade_Numeric']
```

```python
model = LinearRegression()
fs_forward = SequentialFeatureSelector(model,n_features_to_select=3,direction='forward')
```

```python
fs_forward.fit(X, y)
```

```python
forward_features = X.columns[fs_forward.get_support()]
print("Forward Selected Features:")
print(list(forward_features))
```

```python
model = LinearRegression()
fs_backward = SequentialFeatureSelector(model,n_features_to_select=3,direction='backward')
```

```python
fs_backward.fit(X, y)
```

```python
backward_features = X.columns[fs_backward.get_support()]
print("Backward Selected Features:")
print(list(backward_features))
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=3)   # you can change this number
X_pca = pca.fit_transform(X_scaled)

import pandas as pd
X_pca_df = pd.DataFrame(X_pca, columns=['PC1', 'PC2', 'PC3'])

print("PCA Components:")
print(X_pca_df.head())
```
